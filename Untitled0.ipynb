```python
arrA = np.linspace(10,49)
arrB = np.linspace(1,39)
arrC = np.linspace(10,20)
print(f"Array A: {arrA}\n 'array B: '{arrB}\n array C {arrC}.")
soma1 = np.add(arrA,arrB)
mult1 = np.multiply(arrA,arrB)
soma2 = np.add(arrA, arrC)
mult2 = np.multiply(arrB, arrC)
print(f"Soma (A + B): {soma1}\nMult (A * B): {mult1}\nSoma (A + C): {soma2}\nMult (B * C): {mult2}")
```
